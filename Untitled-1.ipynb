```python
folder_path_None = "C:\\Users\\emros\\OneDrive\\Desktop\\No_ACES\\4 - processed"
print(os.path.exists(folder_path_None))
```

```python
channels = [
    "E2","E3","E4","E5","E6","E7","E8","E9","E10","E11",
    "E12","E13","E18","E19","E54","E56","E58","E59","E60"
]
```

```python
# List to collect per-channel averages per file
channel_averages_list = []

# Check folder exists
if os.path.exists(folder_path_None):
    for filename in os.listdir(folder_path_None):
        if filename.endswith(".txt"):
            file_path = os.path.join(folder_path_None, filename)

            # Load file, clean column names
            df_none = pd.read_csv(file_path, sep="\t")  # adjust sep if needed
            df_none.columns = df_none.columns.str.strip()

            # Ensure numeric values
            df_none[channels] = df_none[channels].apply(pd.to_numeric, errors='coerce')

            # Compute mean per channel across time (rows)
            channel_avg_none = df_none[channels].mean()
            channel_avg_none["file"] = filename

            channel_averages_list.append(channel_avg_none)
else:
    print(f"Folder not found: {folder_path_None}")
```

```python
# Combine into DataFrame: rows = files, columns = channels
combined_none = pd.DataFrame(channel_averages_list)
```

```python
# Grand average across all channels and all files
grand_avg_none = combined_none[channels].to_numpy().mean()
print("Grand average activation across all files and channels:", grand_avg_none)
```

```python
# ---- Line Chart ----
mean_per_channel_none = combined_none[channels].mean()  # mean across files per channel

plt.figure(figsize=(12,6))
plt.plot(mean_per_channel_none.index, mean_per_channel_none.values, marker='o', linestyle='-')
plt.xticks(rotation=45)
plt.xlabel("Channel")
plt.ylabel("Average Activation")
plt.title("Average Activation per Channel Across All Files")
plt.grid(True)
plt.tight_layout()
plt.show()
```
